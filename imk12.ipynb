```python
def help_menu():
  print("\n=== HELP DOCUMENTATION ===")
  print("1. nama: Huruf saja")
  print("2. NIM: Angka")
  print("3. Nilai: 0-100")
  print("=============\n")

def input_nama():
  while True:
    nama = input("Masukan nama: ")
    if nama.lower() == "help":
      help_menu()
    elif nama.replace (" ", "").isalpha():
      return nama
    else:
      print("Error: Nama hanya huruf")

def input_nim():
  while True:
    nilai = input("Masukan nim: ")
    if nim.lower() == "help":
      help_menu()
    elif nim.isdigit():
      return nim
    else:
      print("Error: NIM harus angka")

def input_nilai():
  while True:
    nilai = input("Masukan Nilai: ")
    if nilai.lower() == "help":
      help_menu()
      continue
    try:
      nilai = float(nilai)
      if 0 <= nilai <= 100:
        return nilai
      else:
        print("Error: Nilai 0-100")
    except ValueError:
        print("Error: Masukan angka")

print("=== SISTEM INPUT NILAI MAHASISWA ===")
print("Ketik 'help' untuk bantuan\n" )

list_mahasiswa = []
masih = "y"
while masih == "y":
  nama = input_nama()
  nim = input_nim()
  nilai = input_nilai()
  list_mahasiswa.append({
      "nama": nama,
      "nim": nim,
      "nilai": nilai})
  masih = input("Masih mau input? (y/n): ")



print("\n=== Hasil ===")
for i in list_mahasiswa:
  print("Nama :", i['nama'])
  print("NIM :", i['nim'])
  print("Nilai :", i['nilai'])
# print(list_mahasiswa)
```
